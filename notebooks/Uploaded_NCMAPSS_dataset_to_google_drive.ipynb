```python
import zipfile

zip_path = "ncmapss_extracted/17. Turbofan Engine Degradation Simulation Data Set 2/data_set.zip"

with zipfile.ZipFile(zip_path) as z:
    # find the exact entry name inside the archive
    matches = [n for n in z.namelist() if "DS08d" in n]
    print("Found in archive:", matches)

    for name in matches:
        info = z.getinfo(name)
        print(name, "— expected uncompressed size:", info.file_size)

    # extract just this one file fresh, using Python's own unzip logic
    z.extract(matches[0], path="ncmapss_extracted_py")

    # verify its CRC checksum against what the archive expects
    bad_file = z.testzip()
    print("Corrupt entry (None = all good):", bad_file)
```
